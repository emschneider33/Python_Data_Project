```python
import pandas as pd
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
# quick look at key statistics for number columns
df.describe()
```

```python
df['salary_year_avg'].min()
```

```python
# idxmin gives us the index location of where the min value is
min_salary = df['salary_year_avg'].idxmin()
```

```python
# lets investidate the min salary
df.iloc[min_salary]
```

```python
# gives us numerical counts on all the unique values of the job titls short
df['job_title_short'].value_counts()
```

```python
# group the job title short column and lets view the median for each
df.groupby('job_title_short')['salary_year_avg'].median()
```

```python
# example of multiple group by using job title and country
df.groupby(['job_title_short', 'job_country'])['salary_year_avg'].median()
```

```python
# example of group by with multiple value columns
df.groupby('job_title_short')[['salary_year_avg', 'salary_hour_avg']].median()
```

```python
# use .agg operator when wanting to see multiple aggregations
df.groupby('job_title_short')['salary_year_avg'].agg(['min', 'max','median'])
```

**EXERCISE**

```python
# view top 20 countries with their counts
df['job_country'].value_counts().head(20)
```

```python
# filter on us jobs
us_jobs = df[df['job_country'] == "United States"]
```

```python
us_jobs
```

```python
# remove nas in salary year avg column
us_jobs = us_jobs[us_jobs['salary_year_avg'].notna()]
```

```python
# view key statistics for each job title and sort by median
us_jobs.groupby('job_title_short')['salary_year_avg'].agg(['median', 'min', 'max', 'count'] ).sort_values(by= 'median')
```
