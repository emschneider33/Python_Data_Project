```python
import pandas as pd
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

```python
# Lets inspect data
df.info()
```

```python
# we notice job_posted_date is an object
df.job_posted_date[0]
```

```python
# convert to datetime
df['job_posted_date'] = pd.to_datetime(df.job_posted_date)
```

```python
# we can see job_posted_date is now a datetime type, not a string object
df.info()
```

```python
# parse out month and create a new column for just the month
df['job_posted_month'] = df.job_posted_date.dt.month

# check df to make sure new column was added
df.info()
```

```python
# sorting df by date. inplace = True will sort the dataframe permentently wheneevr you call it
df.sort_values(by='job_posted_date', inplace=True)
```

```python
df
```

```python
# dropping salary_hour_avg column (axis =1 drops columns)

df.drop(labels='salary_hour_avg', axis=1, inplace=True)
```

```python
df.info()
```

```python
# filter out na values in salary_year_avg
df.dropna(subset=['salary_year_avg'], inplace=True)
```
