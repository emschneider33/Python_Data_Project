```python
import pandas as pd
from datasets import load_dataset

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()
```

```python
# view first 10 lines of data

df.head(10)

# view last 20 lines of data
# df.tail(20)
```

```python
# specify column names
df['job_title_short']

# specify multiple column names - important to use multiple brackets
df[['job_title_short','job_location']]

# use iloc to specify exact row you'd like to view
df.iloc[90]

# can also use it for ranges
df.iloc[90:101]

# second argument selects the column(s) range
df.iloc[90:101,0:2]
```

```python
#infor provides a lot of valuable information on the dataframe (class,# of rows, # of columns, column names, # of null values, datatype)
df.info()
```

```python
# this will give the statistics summaries of the columns that have numbers
df.describe()
```

```python
#unique function give you all unique values of a column
df.job_title_short.unique()
```

```python
# This gives us a way to filter the dataframe on just those Data Analyst job titles
df[df.job_title_short == 'Data Analyst']

# This gives us a way to filter the dataframe on multiple conditions
df[(df.job_title_short == 'Data Analyst') & (df.salary_year_avg > 100000)]

# returns data analyst jobs where salary is not na
df[(df.job_title_short == 'Data Analyst') & (df.salary_year_avg.notna())]
```
